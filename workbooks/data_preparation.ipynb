```python
import pandas as pd
import holidays
from datetime import datetime, timedelta
```

```python
df = pd.read_csv('../data/combined_data.csv',index_col=None)
```

```python
df
```

```python
df['toll_date'] = pd.to_datetime(df['toll_date'])
df['hour_of_day'] = pd.to_numeric(df['hour_of_day'])
df['minute_of_hour'] = pd.to_numeric(df['minute_of_hour'])
df['crz_entries'] = pd.to_numeric(df['crz_entries'])
```

```python
df.head()
```

```python
us_holidays = holidays.US()

df['holiday_ind'] = df['toll_date'].apply(
    lambda x: x in us_holidays
).astype(int)
```

```python
df['overnight_ind'] = df['time_period'].apply(
    lambda x: 1 if x == 'Overnight' else 0
).astype(int)
```

```python
df['vehicle_class'].value_counts()
```

```python
vehicle_map = {
    'TLC Taxi/FHV':0,
    '1 - Cars, Pickups and Vans':1,
    '2 - Single-Unit Trucks':2,
    '3 - Multi-Unit Trucks':3,
    '4 - Buses':4,
    '5 - Motorcycles':5
}

df['vehicle_category'] = df['vehicle_class'].map(vehicle_map)
```

```python
region_mapping = {
    'Brooklyn':1,
    'Queens':2,
    'New Jersey':3,
    'West Side Highway':4,
    'FDR Drive':5,
    'West 60th St':6,
    'East 60th St':7
}

group_mapping = {
    'Brooklyn Bridge':1,
    'Hugh L. Carey Tunnel':2,
    'Williamsburg Bridge':3,
    'Manhattan Bridge':4,
    'Queensboro Bridge':5,
    'Queens Midtown Tunnel':6,
    'Holland Tunnel':7,
    'Lincoln Tunnel':8,
    'West Side Highway at 60th St':9,
    'FDR Drive at 60th St':10,
    'West 60th St':11,
    'East 60th St':12
}
```

```python
df['region_id'] = df['detection_region'].map(region_mapping).fillna(0).astype(int)
df['group_id'] = df['detection_group'].map(group_mapping).fillna(0).astype(int)
```

```python
df.head(2)
```

```python
df = df.sort_values("toll_10_minute_block")
```

```python
# We want total volume on roadways regardless of whether or not they stayed out of the CRZ
df["traffic_volume"] = (
    df["crz_entries"].fillna(0) +
    df["excluded_roadway_entries"].fillna(0)
)
```

```python
df.drop(['day_of_week','toll_hour','minute_of_hour','hour_of_day','toll_date','crz_entries','excluded_roadway_entries',
         'day_of_week','toll_week','vehicle_class','time_period','detection_group','detection_region'], inplace=True, axis=1)
```

```python
data = df.groupby(
    [
        "toll_10_minute_block",
        "day_of_week_int",
        "holiday_ind",
        "overnight_ind",
        "region_id",
        "group_id"
    ],
    as_index=False
).agg(traffic_volume=("traffic_volume", "sum"))
```

```python
data["toll_10_minute_block"] = pd.to_datetime(
    data["toll_10_minute_block"]
)

data = data.sort_values(
    ["group_id", "toll_10_minute_block"]
).reset_index(drop=True)
```
