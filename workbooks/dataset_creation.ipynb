This combines the original data pull with the appended datapull from August-22 ahead of LSTM modeling

```python
import pandas as pd
```

```python
df1 = pd.read_csv('../data/append_data.csv',index_col=None).drop('Unnamed: 0',axis=1)
```

```python
df2 = pd.read_csv('../data/crz_data_pull_20260720.csv',index_col=None)
```

```python
# All columns from toll_date through detection_region
key_cols = df2.columns[
    df2.columns.get_loc("toll_date"):
    df2.columns.get_loc("detection_region") + 1
].tolist()

# df1 comes second so its values win on duplicate keys
df = (
    pd.concat([df2, df1], ignore_index=True)
      .drop_duplicates(subset=key_cols, keep="last")
      .reset_index(drop=True)
)
```

```python
df.to_csv('../data/combined_data.csv',index=False)
```
